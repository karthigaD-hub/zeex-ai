```python
from ultralytics import YOLO

def train():
    model = YOLO("yolov8s.pt")  # small but accurate

    model.train(
        data="D:\projects\zeex-fire-detect\data\small_detection\data.yaml",
        epochs=80,
        imgsz=640,
        batch=16,
        device=0,          # GPU
        workers=4,
        optimizer="AdamW",
        lr0=0.001,
        cos_lr=True,
        patience=15,
        cache=True,
        mosaic=1.0,
        mixup=0.15,
        hsv_h=0.015,
        hsv_s=0.7,
        hsv_v=0.4,
        degrees=10,
        translate=0.1,
        scale=0.5,
        shear=2,
        flipud=0.1,
        fliplr=0.5
    )

if __name__ == "__main__":
    train()
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  # nano is best for smoke subtlety

model.train(
    data=r"D:\projects\zeex-fire-detect\data\small_detection\data1.yaml",

    imgsz=640,

    epochs=50,            # 🔥 enough
    batch=8,              # ❄️ cool GPU
    patience=10,

    device=0,
    workers=4,

    optimizer="AdamW",
    lr0=0.0006,            # 🔥 lower LR for smoke
    weight_decay=0.0005,

    # 🔥 SMOKE-FRIENDLY AUGMENTATION
    hsv_h=0.015,
    hsv_s=0.25,
    hsv_v=0.20,

    translate=0.05,
    scale=0.30,
    fliplr=0.5,

    mosaic=0.4,            # 🔻 reduce (very important)
    mixup=0.0,             # ❌ never for smoke

    label_smoothing=0.08,  # helps faint objects

    amp=True,
    cache=False,

    project="runs/detect",
    name="smoke_final_v2"
)
```
